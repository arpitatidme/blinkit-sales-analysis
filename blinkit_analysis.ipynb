# Blinkit Sales Analysis — Python EDA
End-to-end exploratory analysis of the Blinkit sales dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
pd.set_option('display.max_columns', None)
```

## Load and inspect data

```python
# Update the filename/path if needed.
df = pd.read_csv('BlinkIT Grocery Data.csv')
print('Shape:', df.shape)
display(df.head())
display(df.describe(include='all').T)
```

## Missing values and cleaning

```python
print(df.isna().sum())
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
if 'item_fat_content' in df.columns:
    df['item_fat_content'] = df['item_fat_content'].replace({'LF':'Low Fat','low fat':'Low Fat','reg':'Regular','regular':'Regular'})
if 'item_weight' in df.columns:
    df['item_weight'] = df['item_weight'].fillna(df['item_weight'].median())
if 'outlet_size' in df.columns:
    df['outlet_size'] = df['outlet_size'].fillna(df['outlet_size'].mode()[0])
print('Shape after cleaning:', df.shape)
```

## Overall KPIs

```python
kpis = {
    'Total Sales': df['sales'].sum(),
    'Average Sales': df['sales'].mean(),
    'Number of Records': len(df),
    'Number of Outlets': df['outlet_identifier'].nunique(),
    'Average Rating': df['rating'].mean()
}
kpis
```

## Sales analysis

```python
item_analysis = df.groupby('item_type')['sales'].agg(total_sales='sum', avg_sales='mean', item_count='count').sort_values('total_sales', ascending=False)
outlet_analysis = df.groupby('outlet_type')['sales'].agg(total_sales='sum', avg_sales='mean', records='count').sort_values('total_sales', ascending=False)
location_analysis = df.groupby('outlet_location_type')['sales'].agg(total_sales='sum', avg_sales='mean', records='count').sort_values('total_sales', ascending=False)
size_analysis = df.groupby('outlet_size')['sales'].agg(total_sales='sum', avg_sales='mean', records='count').sort_values('total_sales', ascending=False)
fat_analysis = df.groupby('item_fat_content')['sales'].agg(total_sales='sum', avg_sales='mean', records='count').sort_values('total_sales', ascending=False)
display(item_analysis)
display(outlet_analysis)
display(location_analysis)
display(size_analysis)
display(fat_analysis)
```

## Correlation analysis

```python
df[['item_visibility','sales','rating']].corr()
```

## Visualizations

```python
plt.figure(figsize=(10,6))
item_analysis['total_sales'].sort_values().plot(kind='barh')
plt.title('Total Sales by Item Type')
plt.xlabel('Sales')
plt.tight_layout()
plt.show()
```

## Key findings
- Fruits & Vegetables generate the highest total sales among item categories.
- Supermarket Type 1 contributes the largest total sales among outlet types.
- Tier 3 outlets generate the highest total sales by location tier.
- Low Fat products contribute more total sales, while Regular products have slightly higher average sales.
- Item Visibility and Rating have very weak linear relationships with Sales.
